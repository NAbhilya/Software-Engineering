```python
import numpy as np
import matplotlib.pyplot as plt

# Quadratic function with different values
def quadratic(x, a=-0.4, b=0.3, c=8):
    return a * x**2 + b * x + c

# Waterfall stages (same names as requested)
stages = [
    "Requirements",
    "Analysis",
    "Design",
    "Implementation",
    "Testing",
    "Maintenance"
]

# X and Y positions (changed range)
x_positions = np.linspace(-3, 3, len(stages))
y_positions = quadratic(x_positions)

# Smooth quadratic curve (different limits)
x_curve = np.linspace(-3.5, 3.5, 500)
y_curve = quadratic(x_curve)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(x_curve, y_curve, linestyle="--", label="Quadratic Curve")

# Waterfall-style horizontal and vertical lines
for i in range(len(stages)):
    if i > 0:
        # Horizontal line
        plt.plot(
            [x_positions[i-1], x_positions[i]],
            [y_positions[i-1], y_positions[i-1]],
            linewidth=2
        )
        # Vertical line
        plt.plot(
            [x_positions[i], x_positions[i]],
            [y_positions[i-1], y_positions[i]],
            linewidth=2
        )

    # Marker and label
    plt.scatter(x_positions[i], y_positions[i], zorder=5)
    plt.text(
        x_positions[i],
        y_positions[i] + 0.7,
        stages[i],
        ha="center",
        fontsize=10
    )

# Styling
plt.title("Waterfall Model Representation using Quadratic Curve")
plt.xlabel("Stages")
plt.ylabel("Value")
plt.ylim(0, max(y_curve) + 3)   # extra top space
plt.grid(True)
plt.legend()
plt.show()
```
