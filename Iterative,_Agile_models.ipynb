```python
import numpy as np
import matplotlib.pyplot as plt

# Quadratic function (represents gradual improvement)
def quadratic(x, a=-0.25, b=1.8, c=22):
    return a * x**2 + b * x + c

# Iterative stages
iterations = [
    "Iteration 1",
    "Iteration 2",
    "Iteration 3",
    "Iteration 4"
]

# X and Y positions
x_positions = np.arange(len(iterations))
y_positions = quadratic(x_positions)

# Smooth quadratic curve
x_curve = np.linspace(-0.5, len(iterations)-0.5, 400)
y_curve = quadratic(x_curve)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(x_curve, y_curve, linestyle="--", label="Quadratic Improvement Curve")

# Plot iterations
for i in range(len(iterations)):
    plt.scatter(x_positions[i], y_positions[i], zorder=5)
    plt.text(
        x_positions[i],
        y_positions[i] + 0.6,
        iterations[i],
        ha="center",
        fontsize=10
    )

plt.title("Iterative Model Representation using Quadratic Curve")
plt.xlabel("Iterations")
plt.ylabel("Predicted Temperature Value")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Quadratic function for Agile development
def quadratic(x, a=-0.2, b=2.2, c=20):
    return a * x**2 + b * x + c

# Agile sprints
sprints = [
    "Sprint 1\n(Data)",
    "Sprint 2\n(Model)",
    "Sprint 3\n(Testing)",
    "Sprint 4\n(Release)"
]

# X and Y positions
x_positions = np.arange(len(sprints))
y_positions = quadratic(x_positions)

# Smooth curve
x_curve = np.linspace(-0.5, len(sprints)-0.5, 400)
y_curve = quadratic(x_curve)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(x_curve, y_curve, linestyle="--", label="Continuous Improvement Curve")

# Plot sprint points
for i in range(len(sprints)):
    plt.scatter(x_positions[i], y_positions[i], zorder=5)
    plt.text(
        x_positions[i],
        y_positions[i] + 0.7,
        sprints[i],
        ha="center",
        fontsize=10
    )

plt.title("Agile Model Representation using Quadratic Curve")
plt.xlabel("Sprints")
plt.ylabel("Predicted Temperature Value")
plt.grid(True)
plt.legend()
plt.show()
```
