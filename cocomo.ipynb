```python
import matplotlib.pyplot as plt

# Constants for different project types (COCOMO Basic Model)
constants = {
    "Organic": (2.4, 1.05),
    "Semi-Detached": (3.0, 1.12),
    "Embedded": (3.6, 1.20)
}

# Function to calculate effort
def cocomo_basic(KLOC, project_type):
    a, b = constants[project_type]
    return a * (KLOC ** b)

# Example: 10 KLOC project
KLOC = 10

# Calculate efforts
efforts = {}
for ptype in constants:
    effort = cocomo_basic(KLOC, ptype)
    efforts[ptype] = effort

# Print results
print("COCOMO Effort Estimation Results:\n")
for ptype, effort in efforts.items():
    print(f"{ptype} Project Effort: {effort:.2f} person-months")

# Plot bar graph
plt.figure()
plt.bar(efforts.keys(), efforts.values())

plt.title(f"COCOMO Effort Estimation for {KLOC} KLOC")
plt.xlabel("Project Type")
plt.ylabel("Effort (Person-Months)")

plt.show()
```

```python
import matplotlib.pyplot as plt

# -----------------------------
# INPUT (from your record)
# -----------------------------
KLOC = 10        # Project size
EAF = 1.15       # Effort Adjustment Factor

# Organic project constants
a = 2.4
b = 1.05

# -----------------------------
# TOTAL EFFORT (Detailed COCOMO)
# -----------------------------
total_effort = a * (KLOC ** b) * EAF

print(f"\nTotal Detailed COCOMO Effort: {total_effort:.2f} person-months\n")

# -----------------------------
# PHASE-WISE DISTRIBUTION
# -----------------------------
phases = [
    "Requirement Analysis",
    "Design",
    "Coding",
    "Testing",
    "Maintenance"
]

# Effort percentages from your document
percentages = [0.10, 0.20, 0.40, 0.20, 0.10]

# Calculate effort for each phase
phase_effort = [total_effort * p for p in percentages]

# Print phase-wise effort
print("Phase-wise Effort Distribution:\n")
for i in range(len(phases)):
    print(f"{phases[i]}: {phase_effort[i]:.2f} person-months")

# -----------------------------
# GRAPH (LIKE YOUR IMAGE)
# -----------------------------
plt.figure(figsize=(8, 5))

plt.bar(phases, phase_effort, color='purple')

plt.title("Effort Distribution in Detailed COCOMO", fontsize=14)
plt.xlabel("Development Phases")
plt.ylabel("Effort (Person-Months)")

# Rotate labels slightly like your image
plt.xticks(rotation=20)

# Grid style
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```
